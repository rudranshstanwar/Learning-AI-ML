### Data Concatenation with Pandas

```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "id" : [1, 2, 3],
    "name" : ["Adam", "Bob", "Eve"],
})

df2 = pd.DataFrame({
    "id" : [4, 5, 6],
    "name" : ["Charlie", "Fester", "Gilian"],
})
```

```python
pd.concat([df1, df2], axis = 0, ignore_index = True)
```

```python
pd.concat([df1, df2], axis = 1, ignore_index = True)
```
