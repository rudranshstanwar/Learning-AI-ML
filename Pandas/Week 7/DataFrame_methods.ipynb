## DataFrame Methods
- df.head(n) - returns the first n rows of the DataFrame
- df.tail(n) - returns the last n rows of the DataFrame
- df.sample(n) - returns a random sample of n rows from the DataFrame
- df.info() - returns a summary of the DataFrame, including the number of non-null values and the data types of each column
- df.shape() - returns the number of rows and columns in the DataFrame in the form of a tuple (rows, columns)
- df.describe() - returns a summary of the numerical columns in the DataFrame, including count, mean, standard deviation, minimum, and maximum values
- df.columns - returns the column names of the DataFrame
- df.nunique() - returns the number of unique values in each column of the DataFrame
