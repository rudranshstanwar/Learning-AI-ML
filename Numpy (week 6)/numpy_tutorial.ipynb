```python
import numpy as np
import time
```

```python
#execution performance
size = 10_000_000
py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"Execution time for list comprehension: {end - start:.4f} seconds")

arr = np.array(py_list)
start = time.time()
sq_array = arr ** 2
end = time.time()
print(f"Execution time for array comprehension: {end - start:.4f} seconds")
```

```python
#memory
import sys
print(f"Memory usage of Python list: {sys.getsizeof(py_list)*len(py_list)} bytes")
print(f"Memory usage of NumPy array: {arr.nbytes} bytes")
```
