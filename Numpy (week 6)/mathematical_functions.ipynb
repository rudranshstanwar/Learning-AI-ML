```python
import numpy as np
```

```python
#aggregate functions
arr = np.array([1, 2, 3, 4, 5])
print(np.sum(arr))
print(np.prod(arr))
print(np.min(arr))
print(np.max(arr))
print(np.argmin(arr))
print(np.argmax(arr))
print(np.mean(arr))
print(np.median(arr))
print(np.var(arr))
```

```python
#power functions
arr = np.array([1, 2, 3, 4, 5])
print(np.square(arr))
print(np.sqrt(arr))
print(np.power(arr, 3))
```

```python
#log
arr = np.array([1, 2, 3, 4, 5])
print(np.log(arr))
print(np.log10(arr))
print(np.exp(arr))
```

```python
#rounding
print(np.round(3.52))
print(np.ceil(3.38))
print(np.floor(3.98))
print(np.trunc(3.98))
```
