## DataFrame

```python
import pandas as pd
import numpy as np
```

```python
#From diction of list
info = {
    "Name" : ["Adam", "Bob", "Eve"],
    "Age" : [23, 24, 25],
    "CGPA" : [9.5, 8.6, 7.2]
}
df = pd.DataFrame(info)
print(df)
print(df.index)
print(df.columns)
```

```python
#From list of lists
df = pd.DataFrame([["Adam", 23], ["Bob", 23], ["Eve", 23]], columns=["Name", "Age"])
print(df)
```

```python
#From numpy array
np_arr = np.array([["Adam", 23], ["Bob", 23], ["Eve", 23]])
df = pd.DataFrame(np_arr, columns=["Name", "Age"])
print(df)
```
