## Series

```python
import pandas as pd
```

```python
s = pd.Series([1, 2, 3, 4, 5])
print(s, type(s))

print(s[1])
```

```python
s = pd.Series([23, 24, 25, 26], index = ["Adam", "Eve", "Bob", "Charlie"])
print(s)

print(s["Eve"])
```

```python
s1 = pd.Series([1, 2, 3, 4])
s2 = pd.Series([5, 6, 7, 8])
print(s1 + s2)
changed_s1 = s1.drop(0)
print(changed_s1)
```
