```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")
df["tax"] = df["income"].apply(lambda x: "20%" if x >= 50_000 else "10%")
gender_map = {"Male" : "M", "Female" : "F"}
df["gender"] = df["gender"].map(gender_map)
df
```

```python
df.columns = ["Id", "Name", "Age", "Country", "Gender", "Income", "Tax"]
df.rename(columns={"Income": "Salary"}, inplace=False)
```

```python
sorted_df = df.sort_values("Income")
sorted_df = sorted_df.reset_index()
sorted_df["Ranking"] = sorted_df["Income"].rank(method = "min")
sorted_df
```
