```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")
```

```python
df.groupby("country")["income"].mean()
```

```python
df.groupby("country")["income"].agg(avg_salary = "mean", min_salary = "min", max_salary =  "max")
```

```python
df.groupby("country").agg(avg_salary = ("income", "mean"), min_salary = ("income", "min"), max_salary =  ("income", "max"))
```
